```python
import numpy as np, random, matplotlib.pyplot as plt
```

```python
def floor(n):
    if n >= 0:
        return int(n)
    return int(n) - 1

N = lambda n: floor(n + 0.5) + (-1) ** (floor(n - floor(n) - 0.5))
```

```python
for i in range(10000):
    n = random.uniform(-20, 20)
    print(abs(n - N(n)), end = ' ')
    if abs(n - N(n)) >= 1 and abs(n - N(n)) <= 1.5:
        print('pass')
    else:
        print('fail, ', n)
        break
print('finished')
```

```python
def f_M(x, M):
    res = 0
    for n in range(M + 1):
        res += (6/7) ** n * np.cos(7 ** n * np.pi *  x)
    return res

def f_M_summand(x, n):
    return (6/7) ** n * np.cos(7 ** n * np.pi *  x)
```

```python
x = np.linspace(-1, 1, 500)
fig, ax = plt.subplots(nrows = 5, ncols = 2, figsize = (20, 40))
c = "orange"
for i in range(5):
    ax[i, 0].plot(x, np.vectorize(f_M)(x, i), label = rf"$M={i}$", color = c)
    ax[i, 0].set_title(rf"$f_{i}=\sum_{{n=0}}^{i}\left(\frac{{6}}{{7}}\right)^n\cos \left(7^n\pi x\right)$")
    ax[i, 1].plot(x, f_M_summand(x,i), label = rf"$n={i}$", color = c)
    ax[i, 1].set_title(rf"$\left(\frac{{6}}{{7}}\right)^{i}\cos \left(7^{i}\pi x\right)$")
plt.tight_layout()
plt.savefig('figures/4_1.png', dpi = 300, bbox_inches = 'tight')
```
